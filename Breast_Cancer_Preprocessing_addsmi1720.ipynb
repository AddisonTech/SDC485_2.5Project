```python
# Import necessary libraries
import pandas as pd
from datetime import datetime
from IPython.display import display, Markdown
import warnings
warnings.filterwarnings("ignore")

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "addsmi1720"

# Load the Breast Cancer dataset into a pandas DataFrame
data = pd.read_csv('breastCancerDataset.csv')

display(Markdown("### Load and Explore the Dataset"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Get a concise summary of the DataFrame
print(data.info())

# Get descriptive statistics for the dataset
display(data.describe())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

display(Markdown("### Preprocess the Data"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Correct any missing values by filling numerical columns with the median
missing_values = data.isnull().sum()
print(f"Missing values before cleaning: {missing_values.sum()}")
for column in data.select_dtypes(include=['number']).columns:
    data[column] = data[column].fillna(data[column].median())
print(f"Missing values after cleaning: {data.isnull().sum().sum()}")

# Identify and remove any duplicate rows
duplicates = data.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")
data = data.drop_duplicates()
print(f"Rows after removing duplicates: {data.shape[0]}")

# Encode the target: Malignant (M) = 1, Benign (B) = 0
data['Diagnosis'] = data['Diagnosis'].map({'M': 1, 'B': 0})

# Standardize the concavity1 and concavity3 features
scaler = StandardScaler()
data[['concavity1', 'concavity3']] = scaler.fit_transform(data[['concavity1', 'concavity3']])
display(data[['concavity1', 'concavity3']].describe().loc[['mean', 'std']].round(3))

# Features and target (ID is a record identifier, not a measurement, so it is excluded)
X = data.drop(columns=['ID', 'Diagnosis'])
y = data['Diagnosis']

# Split the data into training (70%) and testing (30%) sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(f"Training set size: {X_train.shape[0]} samples")
print(f"Test set size: {X_test.shape[0]} samples")
```

```python
from sklearn.svm import SVC

# Fit the SVM model to the training data
svm_model = SVC(random_state=42)
svm_model.fit(X_train, y_train)

# Predict the target values for the test data
y_pred_svm = svm_model.predict(X_test)

print("SVM model training completed.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

display(Markdown("### Evaluate the SVM Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

print(f"Accuracy Score: {accuracy_score(y_test, y_pred_svm):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_svm, target_names=['Benign', 'Malignant']))
```

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid for SVM tuning
svm_param_grid = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf', 'poly']
}

# Use GridSearchCV to find the best hyperparameters
svm_grid = GridSearchCV(SVC(random_state=42), svm_param_grid, cv=5, scoring='accuracy')
svm_grid.fit(X_train, y_train)

# Refit the model using the best parameters
tuned_svm = SVC(**svm_grid.best_params_, random_state=42)
tuned_svm.fit(X_train, y_train)
y_pred_tuned_svm = tuned_svm.predict(X_test)

print("SVM grid search completed.")
```

```python
display(Markdown("### Evaluate the Tuned SVM Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

print(f"Best Parameters: {svm_grid.best_params_}")
print(f"Best Cross-Validation Score: {svm_grid.best_score_:.4f}")
print(f"Accuracy Score: {accuracy_score(y_test, y_pred_tuned_svm):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_tuned_svm, target_names=['Benign', 'Malignant']))
```

```python
from xgboost import XGBClassifier

# Fit the XGBoost model to the training data
xgb_model = XGBClassifier(random_state=42, eval_metric='logloss')
xgb_model.fit(X_train, y_train)

# Predict the target values for the test data
y_pred_xgb = xgb_model.predict(X_test)

print("XGBoost model training completed.")
```

```python
display(Markdown("### Evaluate the XGBoost Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

print(f"Accuracy Score: {accuracy_score(y_test, y_pred_xgb):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_xgb, target_names=['Benign', 'Malignant']))
```

```python
# Define the parameter grid for XGBoost tuning
xgb_param_grid = {
    'n_estimators': [50, 100],
    'learning_rate': [0.1, 0.2],
    'max_depth': [3, 4]
}

# Use GridSearchCV to find the best hyperparameters
xgb_grid = GridSearchCV(XGBClassifier(random_state=42, eval_metric='logloss'), xgb_param_grid, cv=5, scoring='accuracy')
xgb_grid.fit(X_train, y_train)

# Refit the model using the best parameters
tuned_xgb = XGBClassifier(**xgb_grid.best_params_, random_state=42, eval_metric='logloss')
tuned_xgb.fit(X_train, y_train)
y_pred_tuned_xgb = tuned_xgb.predict(X_test)

print("XGBoost grid search completed.")
```

```python
display(Markdown("### Evaluate the Tuned XGBoost Model"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

print(f"Best Parameters: {xgb_grid.best_params_}")
print(f"Best Cross-Validation Score: {xgb_grid.best_score_:.4f}")
print(f"Accuracy Score: {accuracy_score(y_test, y_pred_tuned_xgb):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_tuned_xgb, target_names=['Benign', 'Malignant']))
```

## Analyze and Interpret

**1. In what ways could bias be implemented into this dataset? How can bias be mitigated from similar medical datasets?**

This dataset comes from fine needle aspirate images collected at a single hospital system, so it reflects one patient population, one lab's equipment, and one team's process. The features were measured from cell nuclei that a person selected on the digitized images, so which cells got measured, and how their outlines were traced, depended on human judgment. There are also no age, race, or other demographic fields, so there is no way to check whether the model works equally well for different groups of patients, and patients who were never referred for a biopsy are not in the data at all. The class balance is another built-in choice: 37% of these samples are malignant, which is far higher than in a general screening population, so a model trained here could overcall cancer elsewhere. To reduce bias in similar medical datasets, I would collect data from multiple hospitals and equipment types, record demographic information so performance can be checked for each subgroup, use stratified splits and report recall for the malignant class instead of accuracy alone, since a missed cancer costs far more than a false alarm, and validate the model on an outside dataset before trusting it on new patients.

**2. Compare the performance of the tuned SVM and XGBoost models. Which model performs better and why?**

The tuned SVM performed slightly better on the test set. GridSearchCV picked a linear kernel with C = 10, which reached 97.08% accuracy and a malignant recall of 0.97, missing 2 of the 63 malignant tumors. The tuned XGBoost model (learning rate 0.1, max depth 3, 100 trees) reached 95.91% accuracy and a malignant recall of 0.94, missing 4. The linear kernel won because the features in this dataset, such as radius, perimeter, area, and concavity, separate benign and malignant tumors almost along a straight boundary, so a simple model fits the pattern without overfitting. The RBF and poly kernels scored lower (91% to 92% in cross-validation at best), because only concavity1 and concavity3 were standardized and the unscaled area features dominate their distance calculations, while the linear kernel is less affected by that. The gap between the two tuned models is small, though. XGBoost actually had the higher cross-validation score (96.73% vs 96.23%), and the 1.2 point test difference comes down to 2 patients out of 171, so on a different split the result could flip. Tuning also did not help XGBoost here: the default XGBoost model scored 97.08% on the test set, the same as the tuned SVM, which shows that GridSearchCV chooses settings based on cross-validation and that a small test set can disagree with it.
